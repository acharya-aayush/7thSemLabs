```python
import pandas as pd
import matplotlib.pyplot as plt
import random

# 1. Define your categories
regions = ['Janakpur', 'Kathmandu', 'Lumbini', 'Lalitpur', 'Bhaktapur', 
           'Pokhara', 'Chitwan', 'Mustang', 'Ilam', 'Nagarkot']
months = ['January', 'February', 'March', 'April', 'May', 'June', 
          'July', 'August', 'September', 'October', 'November', 'December']

# 2. Generate the random data
# We use list comprehensions (the [] stuff) because it's faster and cleaner
data = {
    'id': range(1, 1201),
    'region': [random.choice(regions) for _ in range(1200)],
    'month': [random.choice(months) for _ in range(1200)],
    'year': [2025 for _ in range(1200)], # Added Year to be specific
    'no_of_tourists': [random.randint(5, 50) for _ in range(1200)] 
}

df = pd.DataFrame(data)
df.to_csv('nepal_tourist_data.csv', index=False)

# Let's peek at what we made
print(df.head())
```

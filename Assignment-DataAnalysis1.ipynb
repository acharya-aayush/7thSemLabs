```python
region_stats = df.groupby('region')['no_of_tourists'].sum().sort_values(ascending=False)

print("\nTotal Tourists per Region (2023):")
print(region_stats)
```

```python
region_stats.plot(kind='bar', figsize=(10, 5), color='cornflowerblue', edgecolor='black')

plt.title('Annual Tourist Arrivals by Region (2023)')
plt.xlabel('Region')
plt.ylabel('Total Tourists')
plt.xticks(rotation=45) # Rotates the labels (like 'Kathmandu') so they don't overlap
plt.grid(axis='y', linestyle='--', alpha=0.7) # Adds faint lines for easier reading
plt.tight_layout() # Ensures nothing gets cut off
plt.show()
```

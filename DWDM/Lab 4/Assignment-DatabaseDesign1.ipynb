```python
import sqlite3
import pandas as pd

# 1. Connect to a temporary database (runs in memory, fast & easy)
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# 2. Run the SQL Commands (Paste the tables I gave you here)
# We use three quotes ''' to write multi-line SQL
cursor.execute('''
    CREATE TABLE Departments (
        dept_id INT PRIMARY KEY,
        dept_name VARCHAR(50)
    )
''')

cursor.execute('''
    CREATE TABLE Medical_Staff (
        staff_id INT PRIMARY KEY,
        name VARCHAR(100),
        role VARCHAR(50),
        dept_id INT,
        FOREIGN KEY (dept_id) REFERENCES Departments(dept_id)
    )
''')

# (I'll skip creating all 5 tables to keep it short, 
# but you get the idea—you just repeat cursor.execute for each table)

# 3. Insert Data
cursor.execute("INSERT INTO Departments VALUES (1, 'Cardiology')")
cursor.execute("INSERT INTO Medical_Staff VALUES (1, 'Dr. Aayush', 'Doctor', 1)")

# 4. The Final "Select" Query (The cool part)
# We use Pandas to make the output look nice (like a table)
query = "SELECT * FROM Medical_Staff"
df = pd.read_sql_query(query, conn)

print(df)
```
